# 1. Regions

```python
def collecter_regions():
    url = "https://geo.api.gouv.fr/regions"
    response = requests.get(url)
    if response.status_code == 200:
        return response.json()
    return []


regions = collecter_regions()
pprint(regions)
pprint(len(regions))
```

# 3. Communes

```python
import requests


def collecter_departements():
    url = "https://geo.api.gouv.fr/departements"
    r = requests.get(url)
    if r.status_code == 200:
        return [d["code"] for d in r.json()]
    return []

def collecter_communes(code_departement):
    url = f"https://geo.api.gouv.fr/departements/{code_departement}/communes"
    r = requests.get(url)
    if r.status_code == 200:
        return r.json()
    return []

# 1. Récupérer tous les départements
departements = collecter_departements()

# 2. Boucler et afficher 5 communes par département
for dep in departements:
    print(f"\n=== Département {dep} ===")
    communes = collecter_communes(dep)
    for c in communes[:5]:
        print(f"{c['code']} - {c['nom']} (population: {c.get('population', 'N/A')})")
```

# 4. Mairies

```python
import requests
import json

url = "https://api-lannuaire.service-public.gouv.fr/api/explore/v2.1/catalog/datasets/api-lannuaire-administration/exports/json"

# Filtrer uniquement les mairies et télécharger le nombre total de mairies récupérées:

mairies = requests.get(url, params={"where": 'pivot LIKE "mairie"'}).json()

print(f"Nombre total de mairies récupérées : {len(mairies)}\n")

print("Aperçu des 5 premières mairies :\n")

for m in mairies[:5]:
    nom = m.get("nom")
    code_insee = m.get("code_insee_commune")

    # On convertit l'adresse JSON en liste Python
    adresse = json.loads(m["adresse"])

    # On prend la première adresse (si il y a plusieurs adresses, on ne prend que la première)
    adr = adresse[0]

    lat = adr.get("latitude")
    lon = adr.get("longitude")

    print(f"- {nom}")
    print(f"  INSEE : {code_insee}")
    print(f"  Adresse : {adr}")
    print(f"  Latitude : {lat} | Longitude : {lon}\n")
```

# 5. Collèges:

```python
import requests

# --- 1) Récupération des collèges (annuaire) ---
url_annuaire = "https://data.education.gouv.fr/api/explore/v2.1/catalog/datasets/fr-en-annuaire-education/records"

colleges = []
offset = 0
limit = 100

while True:
    params = {
        "limit": limit,
        "offset": offset,
        "where": "type_etablissement = 'Collège'"
    }
    data = requests.get(url_annuaire, params=params).json()
    batch = data.get("results", [])
    if not batch:
        break
    colleges.extend(batch)
    offset += limit

# Indexation par UAI
annuaire_dict = {
    c["identifiant_de_l_etablissement"]: c
    for c in colleges
}

# --- 2) Récupération des effectifs ---
url_effectifs = "https://data.education.gouv.fr/api/explore/v2.1/catalog/datasets/fr-en-college-effectifs-niveau-sexe-lv/records"

effectifs = []
offset = 0

while True:
    params = {"limit": limit, "offset": offset}
    data = requests.get(url_effectifs, params=params).json()
    batch = data.get("results", [])
    if not batch:
        break
    effectifs.extend(batch)
    offset += limit

# Indexation par UAI
effectifs_dict = {
    e["numero_college"]: e["nombre_eleves_total"]
    for e in effectifs
}

# --- 3) Fusion ---
fusion = []

for uai, info in annuaire_dict.items():
    fusion.append({
        "uai": uai,
        "nom": info["nom_etablissement"],
        "adresse": " ".join([
            str(info.get("adresse_1") or ""),
            str(info.get("adresse_2") or ""),
            str(info.get("adresse_3") or "")
        ]).strip(),
        "latitude": info["latitude"],
        "longitude": info["longitude"],
        "code_insee": info["code_commune"],
        "nb_eleves": effectifs_dict.get(uai)  # peut être None si pas trouvé
    })

# --- 4) Afficher 5 résultats fusionnés ---
for c in fusion[:5]:
    print(c)
```

# 6. Lycées:

```python
import requests

HEADERS = {
    "User-Agent": "megabase-projet-formation"
}

TIMEOUT = 30

url = "https://data.education.gouv.fr/api/explore/v2.1/catalog/datasets/fr-en-annuaire-education/records"

all_lycees = []
offset = 0
limit = 100

while True:
    params = {
        "limit": limit,
        "offset": offset,
        "where": 'type_etablissement LIKE "LYCEE%"'
    }

    response = requests.get(url, params=params, headers=HEADERS, timeout=TIMEOUT)
    response.raise_for_status()
    batch = response.json().get("results", [])

    if not batch:
        break  # plus rien à récupérer

    all_lycees.extend(batch)
    offset += limit

print("Nombre total de lycées récupérés :", len(all_lycees))

# Affichage des 5 premiers lycées
for l in all_lycees[:5]:
    uai = l.get("identifiant_de_l_etablissement")
    nom = l.get("nom_etablissement")
    adresse = " ".join([
        str(l.get("adresse_1") or ""),
        str(l.get("adresse_2") or ""),
        str(l.get("adresse_3") or "")
    ]).strip()
    lat = l.get("latitude")
    lon = l.get("longitude")
    code_insee = l.get("code_commune")

    print(f"\nUAI : {uai}")
    print(f"Nom : {nom}")
    print(f"Adresse : {adresse}")
    print(f"Latitude : {lat} | Longitude : {lon}")
    print(f"Code INSEE : {code_insee}")
```

# 7. Gares

```python
import requests

HEADERS = {"User-Agent": "megabase-projet-formation"}
TIMEOUT = 30
LIMIT = 100


# ---------------------------------------------------------
# ÉTAPE 1 — RÉCUPÉRER LES COORDONNÉES (gares-de-voyageurs)
# ---------------------------------------------------------

url_coord = "https://ressources.data.sncf.com/api/explore/v2.1/catalog/datasets/gares-de-voyageurs/records"

coord_data = []
offset = 0

while True:
    params = {"limit": LIMIT, "offset": offset}
    response = requests.get(url_coord, params=params, headers=HEADERS, timeout=TIMEOUT)
    response.raise_for_status()

    batch = response.json().get("results", [])
    if not batch:
        break

    coord_data.extend(batch)
    offset += LIMIT

print("Coordonnées :", len(coord_data), "gares récupérées")

# Indexation par codes_uic (string)
coord_dict = {}

for g in coord_data:
    uic = g.get("codes_uic")  # <-- CHAMP EXACT
    if not uic:
        continue

    coord_dict[uic] = {
        "nom": g.get("nom"),
        "lat": g.get("position_geographique", {}).get("lat"),
        "lon": g.get("position_geographique", {}).get("lon"),
        "code_insee": g.get("codeinsee")
    }


# ---------------------------------------------------------
# ÉTAPE 2 — RÉCUPÉRER LA FRÉQUENTATION (frequentation-gares)
# ---------------------------------------------------------

url_freq = "https://ressources.data.sncf.com/api/explore/v2.1/catalog/datasets/frequentation-gares/records"

freq_data = []
offset = 0

while True:
    params = {"limit": LIMIT, "offset": offset}
    response = requests.get(url_freq, params=params, headers=HEADERS, timeout=TIMEOUT)
    response.raise_for_status()

    batch = response.json().get("results", [])
    if not batch:
        break

    freq_data.extend(batch)
    offset += LIMIT

print("Fréquentation :", len(freq_data), "gares récupérées")


# ---------------------------------------------------------
# ÉTAPE 3 — FUSION DES DEUX DATASETS VIA LE CODE UIC
# ---------------------------------------------------------

fusion = []

for g in freq_data:
    uic = g.get("code_uic_complet")  # champ fréquentation
    if not uic:
        continue

    ref = coord_dict.get(uic)  # correspondance exacte
    if not ref:
        continue

    voyageurs = (
        g.get("total_voyageurs_2024")
        or g.get("total_voyageurs_2023")
        or g.get("total_voyageurs_2022")
    )

    annee = (
        2024 if g.get("total_voyageurs_2024") else
        2023 if g.get("total_voyageurs_2023") else
        2022
    )

    fusion.append({
        "code_uic": uic,
        "nom": ref["nom"],
        "latitude": ref["lat"],
        "longitude": ref["lon"],
        "code_insee": ref["code_insee"],
        "voyageurs_par_an": voyageurs,
        "annee_mesure": annee
    })


# ---------------------------------------------------------
# ÉTAPE 4 — AFFICHAGE
# ---------------------------------------------------------

print("\nNombre total de gares fusionnées :", len(fusion))

for g in fusion[:10]:
    print("\nCode UIC :", g["code_uic"])
    print("Nom :", g["nom"])
    print("Latitude :", g["latitude"], "| Longitude :", g["longitude"])
    print("Code INSEE :", g["code_insee"])
    print("Voyageurs/an :", g["voyageurs_par_an"])
    print("Année mesure :", g["annee_mesure"])
```

# 8. Bibliotheques:

```python
import requests

HEADERS = {"User-Agent": "megabase-projet-formation"}
TIMEOUT = 30
LIMIT = 100

url_bib = "https://data.culture.gouv.fr/api/explore/v2.1/catalog/datasets/adresses-des-bibliotheques-publiques/records"

bibliotheques = []

# Liste des départements (01 à 95 + Corse + DOM-TOM)
departements = (
    [f"{i:02d}" for i in range(1, 96)] +
    ["2A", "2B"] +
    ["971", "972", "973", "974", "975", "976"]
)

for dep in departements:
    offset = 0
    while True:
        params = {
            "limit": LIMIT,
            "offset": offset,
            "where": f"code_departement = '{dep}'"
        }

        response = requests.get(url_bib, params=params, headers=HEADERS, timeout=TIMEOUT)
        response.raise_for_status()

        batch = response.json().get("results", [])
        if not batch:
            break

        bibliotheques.extend(batch)

        offset += LIMIT


print("Bibliothèques :", len(bibliotheques), "bibliothèques récupérées")
```
